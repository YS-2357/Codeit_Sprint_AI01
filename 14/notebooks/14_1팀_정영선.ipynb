# loader.py

### 청킹 전략 실험 결과

- RecursiveCharacterTextSplitter → 청크 수: **1110**
- TokenTextSplitter → 청크 수: **5230**

#### 분석
- Recursive 방식은 문장을 문자 수 기준으로 나누며, 전체적으로 큰 덩어리를 형성한다.
- Token 방식은 LLM 입력 단위에 가까운 토큰 기준으로 분할하여 더 많은 청크를 생성한다.

#### 결론
- Token 청킹은 문맥 정밀도와 검색 정확도에 더 유리할 수 있지만, 속도 및 메모리 측면에서 비효율적일 수 있다.
- 실제 질의응답 결과의 품질을 비교해 전략을 선택하는 것이 바람직하다.

```python
def load_and_clean_pdf(file_path):
    """
    PDF 파일을 로드하고 페이지별 텍스트를 정제하여 반환합니다.

    Args:
        file_path (str): 로컬 PDF 파일 경로

    Returns:
        List[Document]: 정제된 LangChain Document 객체 리스트
    """
    from langchain_community.document_loaders import PyPDFLoader
    import re

    def clean_text(text):
        text = re.sub(r"[\ue000-\uf8ff]", "", text)  # 이상 유니코드 제거
        text = re.sub(r"(?<!\n)\n(?!\n)", " ", text)  # 줄바꿈 하나는 띄어쓰기로 치환
        text = re.sub(r"[·•・∙ㆍ]+", " ", text)        # 불필요한 점 제거
        return text.strip()

    loader = PyPDFLoader(file_path)
    pages = loader.load()

    for page in pages:
        page.page_content = clean_text(page.page_content)

    return pages


def split_documents(pages, strategy="recursive", chunk_size=500, chunk_overlap=100):
    """
    문서 페이지를 청크로 분할합니다. 전략에 따라 문자 기반 또는 토큰 기반 분할이 가능합니다.

    Args:
        pages (List[Document]): LangChain Document 객체 리스트
        strategy (str): "recursive" 또는 "token" 청킹 전략
        chunk_size (int): 청크 최대 길이
        chunk_overlap (int): 청크 간 중첩 길이

    Returns:
        List[Document]: 청크 단위로 분할된 문서 리스트
    """
    from langchain_text_splitters import (
        RecursiveCharacterTextSplitter,
        TokenTextSplitter,
    )

    if strategy == "recursive":
        splitter = RecursiveCharacterTextSplitter(
            chunk_size=chunk_size,
            chunk_overlap=chunk_overlap
        )
    elif strategy == "token":
        splitter = TokenTextSplitter(
            chunk_size=chunk_size,
            chunk_overlap=chunk_overlap,
            model_name="gpt2"  # 기본 tokenizer
        )
    else:
        raise ValueError("Unknown strategy: choose 'recursive' or 'token'")

    return splitter.split_documents(pages)
```

```python
loader_config = {
    "file_path": "/content/drive/MyDrive/Colab Notebooks/25-1학기/25-18~19주차/2024년+원천징수의무자를+위한+연말정산+신고안내.pdf",
    "chunk_size": 500,
    "chunk_overlap": 100,
    "strategy": "recursive",
}
```

```python
# 문서 불러오기
pages = load_and_clean_pdf(loader_config["file_path"])
```

```python
# 청킹 실험
splits_recursive = split_documents(pages, strategy="recursive", chunk_size=500, chunk_overlap=100)
splits_token = split_documents(pages, strategy="token", chunk_size=200, chunk_overlap=50)

# 길이 확인
print("Recursive:", len(splits_recursive))
print("Token:", len(splits_token))
```

# embedder.py

### 임베딩 생성 및 벡터 데이터베이스 구성

- 청킹된 문서 조각(`splits_recursive`, `splits_token`)에 대해 Hugging Face의 `nlpai-lab/KoE5` 임베딩 모델을 사용해 벡터를 생성하였습니다.
- nlpai-lab/KoE5는 한국어 문서의 의미 표현에 적합한 사전학습 임베딩 모델로, 검색 정확도 향상에 효과적입니다.
- 생성된 벡터는 LangChain에서 제공하는 FAISS 벡터 저장소에 저장하였고, 이후 검색 가능하도록 retriever 객체로 변환하였습니다.

```python
import faiss
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.docstore.in_memory import InMemoryDocstore
from langchain_community.vectorstores import FAISS

def build_vector_store(splits, embedding_model_name="nlpai-lab/KoE5"):
    """
    문서 청크에 대한 임베딩을 생성하고 FAISS 벡터 데이터베이스에 저장하여 retriever를 반환합니다.

    Args:
        splits (List[Document]): 청크로 분할된 문서 리스트
        embedding_model_name (str): Hugging Face의 사전학습 임베딩 모델 이름

    Returns:
        VectorStoreRetriever: 검색을 수행할 수 있는 LangChain retriever 객체
    """
    embeddings = HuggingFaceEmbeddings(model_name=embedding_model_name)
    dim = len(embeddings.embed_query("hello world"))  # 임베딩 차원 수 추론
    index = faiss.IndexFlatL2(dim)

    vector_store = FAISS(
        embedding_function=embeddings,
        index=index,
        docstore=InMemoryDocstore(),
        index_to_docstore_id={},
    )
    vector_store.add_documents(splits)
    return vector_store.as_retriever()
```

```python
retriever_recursive = build_vector_store(splits_recursive, embedding_model_name="nlpai-lab/KoE5")
retriever_token = build_vector_store(splits_token, embedding_model_name="nlpai-lab/KoE5")
```

# llm_setup.py

### 언어 모델 및 토크나이저 설정

- 텍스트 생성을 위한 LLM으로 Hugging Face의 `beomi/KoAlpaca-Polyglot-5.8B`를 사용하였습니다. 이 모델은 한국어 문서에 최적화되어 있어, 한국어 질문-응답 시스템에 적합합니다.
- GPU 환경 제약이 있는 상황을 고려하여 4bit 양자화를 적용하였고, 비GPU 환경에서도 실행 가능하도록 CPU fallback 로직을 포함했습니다.
- 생성 품질 조절을 위해 `temperature`, `repetition_penalty`, `max_new_tokens` 등의 하이퍼파라미터를 외부 config를 통해 전달하여 조절 가능하게 구성하였습니다.

```python
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig, pipeline
from langchain_huggingface import HuggingFacePipeline
import torch

def load_llm(model_name: str, generation_config: dict, quant_config: dict):
    """
    Hugging Face LLM을 로드하고 LangChain의 HuggingFacePipeline 형태로 반환합니다.
    chat_template 미지원 모델 대응을 위해 ChatHuggingFace 대신 HuggingFacePipeline을 사용합니다.

    Args:
        model_name (str): Hugging Face 모델 이름
        generation_config (dict): 텍스트 생성 설정
        quant_config (dict): 양자화 설정

    Returns:
        HuggingFacePipeline: LangChain 호환 LLM 파이프라인 객체
    """
    if torch.cuda.is_available():
        bnb_config = BitsAndBytesConfig(
            load_in_4bit=quant_config['load_in_4bit'],
            bnb_4bit_use_double_quant=quant_config['use_double_quant'],
            bnb_4bit_quant_type=quant_config['quant_type'],
            bnb_4bit_compute_dtype=getattr(torch, quant_config['compute_dtype']),
            llm_int8_enable_fp32_cpu_offload=quant_config['enable_fp32_offload']
        )
        model = AutoModelForCausalLM.from_pretrained(
            model_name,
            quantization_config=bnb_config,
            trust_remote_code=True,
        )
    else:
        model = AutoModelForCausalLM.from_pretrained(
            model_name,
            torch_dtype=torch.float32,
            trust_remote_code=True,
        )

    tokenizer = AutoTokenizer.from_pretrained(model_name)

    pipe = pipeline(
        model=model,
        tokenizer=tokenizer,
        task="text-generation",
        do_sample=True,
        temperature=generation_config["temperature"],
        repetition_penalty=generation_config["repetition_penalty"],
        return_full_text=False,
        max_new_tokens=generation_config["max_new_tokens"],
    )

    return HuggingFacePipeline(pipeline=pipe)
```

```python
quant_config = {
    "load_in_4bit": True,
    "use_double_quant": True,
    "quant_type": "nf4",
    "compute_dtype": "float16",
    "enable_fp32_offload": True,
}

generation_config = {
    "temperature": 0.1,
    "repetition_penalty": 1.2,
    "max_new_tokens": 1000,
}

language_model_name = "beomi/KoAlpaca-Polyglot-5.8B"

chat_model = load_llm(language_model_name, generation_config, quant_config)
```

# rag_chain.py

### RAG 시스템 구현

- 사용자의 질문을 받아 문서 청크 중 관련된 부분을 `retriever`를 통해 찾아낸 후, 이를 LLM의 프롬프트 맥락으로 주입해 응답을 생성하는 Retrieval-Augmented Generation(RAG) 시스템을 구현하였습니다.
- 청킹 전략(Recursive vs Token)에 따라 `retriever`를 따로 구성하고, 동일한 `prompt_template` 및 LLM으로 각각 실험하였습니다.

```python
from langchain.schema.runnable import RunnablePassthrough
from langchain.schema.output_parser import StrOutputParser
from langchain_core.prompts import PromptTemplate

def format_docs(docs):
    """
    검색된 문서 목록을 단일 문자열로 합칩니다.

    Args:
        docs (List[Document]): 검색된 문서 객체 리스트

    Returns:
        str: 개별 문서의 page_content를 개행으로 구분하여 이어 붙인 문자열
    """
    return "\n\n".join(doc.page_content for doc in docs)

def build_rag_chain(retriever, llm, prompt_template: str, top_k: int = 3):
    """
    검색 기반 생성(RAG) 체인을 구성합니다. 사용자의 질문을 받아,
    관련 문서를 검색하고 해당 문서를 맥락으로 LLM이 응답을 생성합니다.

    Args:
        retriever: 문서 검색을 위한 retriever 객체 (VectorStoreRetriever 등)
        llm: LangChain의 LLM 객체 (예: ChatHuggingFace)
        prompt_template (str): context + question을 포함한 프롬프트 템플릿 문자열
        top_k (int, optional): 검색할 문서 수 (기본값은 3)

    Returns:
        Runnable: LangChain의 체인 구성 객체. `.invoke(question)`으로 응답 생성 가능
    """
    retriever.search_kwargs = {"k": top_k}
    prompt = PromptTemplate.from_template(prompt_template)

    return (
        {"context": retriever | format_docs, "question": RunnablePassthrough()}
        | prompt
        | llm
        | StrOutputParser()
    )
```

```python
# 청킹 전략별로 RAG 체인 생성
rag_chain_recursive = build_rag_chain(retriever_recursive, chat_model, prompt_template)
rag_chain_token = build_rag_chain(retriever_token, chat_model, prompt_template)
```

# main.py

```python
questions = [
    "연말 정산 때 비거주자가 주의할 점을 알려 줘.",
    "2024년 개정 세법 중에 월세와 관련한 내용이 있을까?",
    "기부금 공제는 어떤 기준으로 적용되나요?"
]

for q in questions:
    print(f"[질문] {q}")
    print("\n[Recursive 기반 응답]")
    print(rag_chain_recursive.invoke(q))
    print("\n[Token 기반 응답]")
    print(rag_chain_token.invoke(q))
    print("="*80)
```
