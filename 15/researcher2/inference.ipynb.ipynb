```python
import pandas as pd
import pickle
import os
import sklearn
import numpy
```

```python
MODEL_PATH = "../shared/model.pkl"
TEST_PATH = "../shared/test.csv"
RESULT_PATH = "../shared/result.csv"
```

```python
with open(MODEL_PATH, 'rb') as f:
    ct, model = pickle.load(f)
print(ct, model)
```

```python
test_df = pd.read_csv(TEST_PATH)
test_df.head()
```

```python
X_data = ct.transform(test_df)
X_data
```

```python
feature_names = ct.get_feature_names_out()
X_trans = pd.DataFrame(X_data, columns=feature_names, index=test_df.index)
```

```python
X_trans
```

```python
y_pred = model.predict(X_trans)
y_pred
```

```python
result_df = pd.DataFrame(y_pred)
result_df.to_csv(RESULT_PATH, index=False)
```
